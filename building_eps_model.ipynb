```python
#checking presence of metabolites
new_metabolites=[]
metabolite= ["udcpp_w_c","uacgam_c", "udpacgal_c","udpgal_c","udp_c", "pyr_c", "ump_c","glu__L_c", "akg_c","accoa_c", "ump_c", "cpd24246_c", "cpd02810_c", "ukquivo_c", "udcppbac_c", "udcppbacglcnacgaln_c", "udcppbacglcnacgalngal_c","eps_c"]
for metabolite in metabolite:
    if metabolite in model_bs.metabolites:
        print(f"Metabolite '{metabolite}' is present in the model.")
    else:
        print(f"Metabolite '{metabolite}' is not present in the model.")
        new_metabolites.append(metabolite)
print(new_metabolites)
```

```python
#adding the new metabolites
new_metabolites={'cpd24246_c':'C19H30N4O16P2', 'cpd02810_c':'C17H28N4O15P2', 'ukquivo_c':'C18H30N4O16P2', 'udcppbac_c':'', 'udcppbacglcnacgaln_c':'', 'udcppbacglcnacgalngal_c':'', 'eps_c':'','udcppbacglcnac_c':''}
for metabolite, formula in new_metabolites.items():
    if metabolite not in model_bs.metabolites:
        print(f"Adding missing metabolite: {metabolite} with formula '{formula}'")
        model_bs.add_metabolites([cobra.Metabolite(metabolite, compartment='c', formula=formula)])
```

```python
#Adding the reactions
def add_rxn(model, rid, stoich, name, lb=0, ub=1000):
    rxn = cobra.Reaction(rid)

    rxn.name = name 

    rxn.lower_bound = lb
    rxn.upper_bound = ub

    rxn.add_metabolites(
        {
            model.metabolites.get_by_id(k):v
            for k,v in stoich.items()
        }
    )

    model.add_reactions([rxn])
```

```python

# UDP-GlcNAc -> UDP-4-keto sugar
add_rxn(
model_bs,
'UDP_GlcNAc_dehydrogenase',
{"uacgam_c":-1,
"nad_c":-1,
"ukquivo_c":1,
"h2o_c":1,
"nadh_c":1},
'UDP-GlcNAc oxidation'
)



# UDP-4-keto -> UDP-Bacillosamine
# KEGG R04529

add_rxn(
model_bs,
"UDP_bacillosamine_synthase",
{
"ukquivo_c":-1,
"akg_c":-1,
"cpd02810_c":1,
"glu__L_c":1
},
"UDP-bacillosamine formation"
)



# Acetylation
# KEGG R10099               

add_rxn(
model_bs,
"UDP_diacetylbacillosamine_synthase",
{
"cpd02810_c":-1,
"accoa_c":-1,
"cpd24246_c":1,
"coa_c":1
},
"UDP-N,N'-diacetylbacillosamine formation"
)
```

```python
# RXN1
# udcpp + udpbac -> udcppbac + ump

add_rxn(
    model_bs,
    "UDCPPBACT",
    {
        "udcpp_w_c": -1,
        "cpd24246_c": -1,
        "udcppbac_c": 1,
        "ump_c": 1
    },
    "UDP-Bac transfer to undecaprenyl phosphate"
)



# RXN2
# udcppbac + udpglcnac -> udcppbacglcnac + udp

add_rxn(
    model_bs,
    "BACGLCNACT",
    {
        "udcppbac_c": -1,
        "uacgam_c": -1,
        "udcppbacglcnac_c": 1,
        "udp_c": 1
    },
    "GlcNAc transfer"
)



# RXN3
# udcppbacglcnac + udpgaln -> udcppbacglcnacgaln + udp

add_rxn(
    model_bs,
    "BACGALNACT",
    {
        "udcppbacglcnac_c": -1,
        "udpacgal_c": -1,
        "udcppbacglcnacgaln_c": 1,
        "udp_c": 1
    },
    "GalNAc transfer"
)



# RXN4
# udcppbacglcnacgaln + udpgal
# -> udcppbacglcnacgalngal + udp

add_rxn(
    model_bs,
    "BACGALT",
    {
        "udcppbacglcnacgaln_c": -1,
        "udpgal_c": -1,
        "udcppbacglcnacgalngal_c": 1,
        "udp_c": 1
    },
    "Galactose transfer"
)



# RXN5
# udcppbacglcnacgalngal + 2 udpgal + 2 pyr
# -> eps + 2 udp + udcpp

add_rxn(
    model_bs,
    "EPSSYNTH",
    {
        "udcppbacglcnacgalngal_c": -1,
        "udpgal_c": -2,
        "pyr_c": -2,

        "eps_c": 1,
        "udp_c": 2,
        "udcpp_w_c": 1
    },
    "EPS synthesis and lipid carrier recycling"
)



```

```python
dm_eps = add_rxn(model_bs, "DM_eps_c", {"eps_c": -1}, "EPS demand / export")
 
 
# ─────────────────────────────────────────────────────────────────────────────
# 5.  FBA SANITY CHECK
# ─────────────────────────────────────────────────────────────────────────────
print("\n[4/4] FBA sanity check — maximise EPS production")
print("─" * 60)
 
pathway_rxn_ids = [
    "UDP_GlcNAc_dehydrogenase",
    "UDP_bacillosamine_synthase",
    "UDP_diacetylbacillosamine_synthase",
    "UDCPPBACT",
    "BACGLCNACT",
    "BACGALNACT",
    "BACGALT",
    "EPSSYNTH",
    "dm_eps",
]
 
dm_eps = model_bs.reactions.get_by_id("DM_eps_c")
with model_bs:
    model_bs.objective = dm_eps
    solution = model_bs.optimize()
 
    print(f"Status        : {solution.status}")
    print(f"EPS objective : {solution.objective_value:.6f}")
 
    if solution.status == "optimal" and solution.objective_value > 0:
        print("\n✓ EPS can be produced — pathway is connected.\n")
        print(f"{'Reaction':<38} Flux")
        print("─" * 50)
        for rid in pathway_rxn_ids:
            flux = solution.fluxes.get(rid, 0.0)
            print(f"{rid:<38} {flux: .4f}")
    else:
        print("\n✗ EPS flux = 0. Diagnosing which step is blocked...\n")
        # Quick per-reaction diagnostic: maximise each pathway reaction's own
        # flux individually (ignoring EPS objective) to see where the chain
        # breaks — a reaction stuck at 0 here means an upstream precursor
        # is unavailable under the current medium/bounds.
        print(f"{'Reaction':<38} Max individual flux")
        print("─" * 60)
        for rid in pathway_rxn_ids:
            with model_bs:
                model_bs.objective = model_bs.reactions.get_by_id(rid)
                test_sol = model_bs.optimize()
                val = test_sol.objective_value if test_sol.status == "optimal" else float("nan")
                flag = "  ← blocked" if (val is None or val == 0) else ""
                print(f"{rid:<38} {val: .4f}{flag}")
        print("\nTip: a reaction blocked here usually means one of its substrates "
              "(akg_c, accoa_c, uacgam_c, udpacgal_c, udpgal_c, etc.) has no supply "
              "route under the current medium — check with, e.g.:")
        print("  model_bs.metabolites.get_by_id('akg_c').summary()")

 
 
```

```python
"""
add_eps_gpr.py
==============
Adds Gene-Protein-Reaction (GPR) rules to the EPS biosynthesis reactions
in the B. subtilis iBS1147 genome-scale model.

Gene assignments are based on:
  - SubtiWiki (https://subtiwiki.uni-goettingen.de)
  - Roux et al. (2015) PNAS — B. subtilis EPS structure
  - Kearns et al. (2005) Mol Microbiol — eps operon characterisation
  - KEGG reaction entries R04529, R10099


NOTE: Run Section 1 FIRST to confirm whether your model uses BSU locus
      tags or gene names, then edit USE_LOCUS_TAGS accordingly.
"""

import cobra

# ─────────────────────────────────────────────────────────────────────────────
# SECTION 0 — GENE ID FORMAT
# Set USE_LOCUS_TAGS = True  → uses BSU26490, BSU26500 ...
# Set USE_LOCUS_TAGS = False → uses epsD, epsE, epsF ...
# ─────────────────────────────────────────────────────────────────────────────
USE_LOCUS_TAGS = True

GENE_IDS = {
    "epsC": "BSU34350" if USE_LOCUS_TAGS else "epsC",
    "epsL": "BSU34250" if USE_LOCUS_TAGS else "epsL",
    "epsD": "BSU34340" if USE_LOCUS_TAGS else "epsD",
    "epsE": "BSU34330" if USE_LOCUS_TAGS else "epsE",
    "epsF": "BSU34320" if USE_LOCUS_TAGS else "epsF",
    "epsH": "BSU34300" if USE_LOCUS_TAGS else "epsH",
    "epsI": "BSU34290" if USE_LOCUS_TAGS else "epsI",
    "epsJ": "BSU34280" if USE_LOCUS_TAGS else "epsJ",
    "epsK": "BSU34265" if USE_LOCUS_TAGS else "epsK",
    "epsL": "BSU34250" if USE_LOCUS_TAGS else "epsL",
    "epsM": "BSU34240" if USE_LOCUS_TAGS else "epsM",
    "epsN": "BSU34230" if USE_LOCUS_TAGS else "epsN",
    "epsO": "BSU34220" if USE_LOCUS_TAGS else "epsO",
}

g = GENE_IDS  # shorthand

# ─────────────────────────────────────────────────────────────────────────────
# SECTION 1 — GENE ID CHECKER
# ─────────────────────────────────────────────────────────────────────────────
print("=" * 65)
print(" SECTION 1 — GENE ID CHECKER")
print("=" * 65)

model_gene_ids = {gene.id for gene in model_bs.genes}
print(f"Total genes in model: {len(model_gene_ids)}\n")
print(f"  {'Gene role':<10} {'ID to assign':<15} {'In model?'}")
print("  " + "─" * 38)

all_found = True
for role, gene_id in g.items():
    found = gene_id in model_gene_ids
    flag  = "✓" if found else "✗  ← NOT FOUND"
    if not found:
        all_found = False
    print(f"  {role:<10} {gene_id:<15} {flag}")

if not all_found:
    print("\n⚠  Some gene IDs not found. Try:")
    print("   [g.id for g in model_bs.genes if 'eps' in g.id.lower()]")
    print("   [g.id for g in model_bs.genes if 'BSU265' in g.id]")
    print("   Then flip USE_LOCUS_TAGS or edit GENE_IDS manually.")
else:
    print("\n✓  All gene IDs confirmed in model — safe to proceed.")

# ─────────────────────────────────────────────────────────────────────────────
# SECTION 2 — GPR MAP (reaction → rule + description)
#
# Confidence levels:
#   HIGH   = direct experimental evidence in B. subtilis
#   MEDIUM = inferred from C. jejuni pgl pathway homology
# ─────────────────────────────────────────────────────────────────────────────

GPR_MAP = {

    # uacgam_c + nad_c → ukquivo_c + h2o_c + nadh_c
    "UDP_GlcNAc_dehydrogenase": (
        g["epsC"],
        "EpsC — Probable polysaccharide biosynthesis protein EpsC"
    ),

    # ukquivo_c + akg_c → cpd02810_c + glu__L_c   KEGG R04529
    "UDP_bacillosamine_synthase": (
        g["epsN"],
        "EpsN — Putative pyridoxal phosphate-dependent aminotransferase EpsN"
    ),

    # cpd02810_c + accoa_c → cpd24246_c + coa_c   KEGG R10099
    "UDP_diacetylbacillosamine_synthase": (
        g["epsM"],
        "EpsM — UDP-N-acetylbacillosamine N-acetyltransferase"
    ),

    # udcpp_w_c + cpd24246_c → udcppbac_c + ump_c
    # Transfers Bac-1-P to Und-PP, displacing UMP  [HIGH]
    "UDCPPBACT": (
        g["epsL"],
        "EpsL — Uncharacterized sugar transferase EpsL"
    ),

    # udcppbac_c + uacgam_c → udcppbacglcnac_c + udp_c
    "BACGLCNACT": (
        g["epsD"],
        "EpsD — Putative glycosyltransferase"
    ),

    # udcppbacglcnac_c + udpacgal_c → udcppbacglcnacgaln_c + udp_c
    "BACGALNACT": (
        g["epsE"],
        "EpsE — Putative glycosyltransferase"
    ),

    # udcppbacglcnacgaln_c + udpgal_c → udcppbacglcnacgalngal_c + udp_c
    "BACGALT": (
        g["epsF"],
        "EpsF — Putative glycosyltransferase"
    ),

    # udcppbacglcnacgalngal_c + 2 udpgal_c + 2 pyr_c
    #   → eps_c + 2 udp_c + udcpp_w_c
    # EpsI–EpsM: polymerase complex; all 5 subunits required → AND rule  [MEDIUM]
    "EPSSYNTH": (
        f"{g['epsI']} and {g['epsJ']} and {g['epsH']} and {g['epsO']} ",
        "EpsI-EpsJ-EpsH-EpsO — EPS polymerase/pyruvyltransferase complex"
    ),

    # Artificial sink — no enzyme
    "DM_eps_c": (
        "",
        "EPS demand reaction (no gene — artificial export sink)"
    ),
}

# ─────────────────────────────────────────────────────────────────────────────
# SECTION 3 — APPLY GPR RULES
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "=" * 65)
print(" SECTION 2 — APPLYING GPR RULES")
print("=" * 65)

success, not_found = [], []

for rxn_id, (rule, description) in GPR_MAP.items():
    if rxn_id not in model_bs.reactions:
        print(f"  ✗ '{rxn_id}' not in model — skipping.")
        not_found.append(rxn_id)
        continue
    rxn = model_bs.reactions.get_by_id(rxn_id)
    rxn.gene_reaction_rule = rule
    print(f"  ✓ {rxn_id}")
    print(f"      {description}")
    print(f"      GPR : {rule if rule else '(none)'}\n")
    success.append(rxn_id)

print(f"  {len(success)} reactions updated | {len(not_found)} not found")

# ─────────────────────────────────────────────────────────────────────────────
# SECTION 4 — VERIFICATION TABLE
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "=" * 65)
print(" SECTION 3 — VERIFICATION TABLE")
print("=" * 65)
print(f"  {'Reaction':<38} {'Genes linked'}")
print("  " + "─" * 60)

new_eps_genes = set()
for rxn_id in GPR_MAP:
    if rxn_id not in model_bs.reactions:
        continue
    rxn   = model_bs.reactions.get_by_id(rxn_id)
    genes = [g_obj.id for g_obj in rxn.genes]
    new_eps_genes.update(genes)
    print(f"  {rxn_id:<38} {', '.join(genes) if genes else '—'}")

print(f"\n  Unique genes linked to EPS reactions: {len(new_eps_genes)}")
for gene_id in sorted(new_eps_genes):
    gene   = model_bs.genes.get_by_id(gene_id)
    n_rxns = len(gene.reactions)
    print(f"    {gene_id:<15} participates in {n_rxns} reaction(s)")

print(f"\n  Total genes in model: {len(model_bs.genes)}")

# ─────────────────────────────────────────────────────────────────────────────
# SAVE
# ─────────────────────────────────────────────────────────────────────────────
cobra.io.write_sbml_model(model_bs, "ibsu1147_with_eps.xml")
print("\nSaved → ibsu1147_with_eps.xml")
```
